```python
from pathlib import Path

import pandas as pd

MOVEMENT_DIR = Path(
    r"C:\Users\Sheel\OneDrive\Documents\Default Project\data\pads-parkinsons-disease-smartwatch-dataset-1.0.0\pads-parkinsons-disease-smartwatch-dataset-1.0.0\preprocessed\movement"
)

file_list = pd.read_csv(MOVEMENT_DIR.parent / "file_list.csv")
```

```python
groups = {
    "Atypical Parkinsonism": [15, 18, 32, 64, 106, 110, 114, 118, 158, 168, 238, 280, 291, 304, 397],
    "Healthy": [46, 78, 81, 108, 148, 153, 160, 165, 214, 254, 295, 319, 335, 389, 463],
    "Essential Tremor": [66, 70, 72, 91, 112, 124, 143, 200, 275, 278, 302, 340, 350, 386, 396],
    "Parkinson's": [4, 95, 109, 113, 175, 206, 226, 249, 345, 347, 362, 365, 395, 406, 442],
}

id_to_group = {pid: g for g, pids in groups.items() for pid in pids}
subset = file_list[file_list["id"].isin(id_to_group)].set_index("id")
subset["group"] = [id_to_group[i] for i in subset.index]
print(len(subset), "participants total")
print(subset.groupby("group").size().to_string())
```

```python
summary = subset.groupby("group").agg(
    total_participants=("age", "size"),
    age_min=("age", "min"),
    age_max=("age", "max"),
    mean_age=("age", "mean"),
    stdev_age=("age", "std"),
    males=("gender", lambda s: (s == "male").sum()),
    females=("gender", lambda s: (s == "female").sum()),
)
summary.to_string()
print(summary)
```

```python
for g, pids in groups.items():
    sub = subset.loc[pids]
    print(f"=== {g} ===")
    print(f"Total participants: {len(sub)}")
    print(f"Age range: {sub['age'].min()} - {sub['age'].max()}")
    print(f"Mean age: {sub['age'].mean()}")
    print(f"STDEV of age: {sub['age'].std()}")
    print(f"Males: {(sub['gender'] == 'male').sum()}")
    print(f"Females: {(sub['gender'] == 'female').sum()}")
    print()
```
